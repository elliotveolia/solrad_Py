Wildfire Impact on CAISO

Each figure corresponds to a significant wildfire event, clearly marked on the visualization. 
The plots display both the day-ahead forecasted load and actual observed load for CAISO, 
spanning the period surrounding each wildfire occurrence. This comparison reveals the extent 
to which smoke and atmospheric conditions affected grid demand patterns during these events.
